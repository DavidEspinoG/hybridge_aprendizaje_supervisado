# Análisis Exploratorio de Datos (EDA)

Exploraremos los datos limpios para comprender las relaciones entre las variables de supervivenia.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv('data/titanic_clean.csv')
df.head()
```

# Supervivencia por Género

```python
plt.figure(figsize=(8,5))
sns.barplot(x='Sex', y='Survived', data=df, palette='viridis')
plt.title('Probabilidad de Supervivencia por Género')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.barplot(x='Pclass', y= 'Survived', data=df, palette='magma')
plt.title('Probabilidad de Supervivencia por Clase')
plt.show()
```

# Distribución de Edades

```python
plt.figure(figsize=(10,6))
sns.histplot(data=df, x='Age', hue= 'Survived', kde=True, element='step')
plt.title('Distribución de Edad por Superviviencia')
plt.show()
```

```python
df['Sex'] = df['Sex'].map({'male': 0, 'female':1})

df = pd.get_dummies(df, columns=['Embarked'], prefix='Emb')

df.head()
```

```python
df.to_csv('data/titanic_feature.csv', index= False)

print("Datos guardados en data/titanic_feature.csv")
```

# Cantidad de supervivientes por  clase 


```python
survivors_df = df[df["Survived"] == 1]
# 2. Set a clean visual style
sns.set_theme(style="whitegrid")

# 3. Create a count plot for each passenger class using the filtered data
ax = sns.countplot(x='Pclass', data=survivors_df, palette='Set1')

# 4. Add the exact numbers on top of each bar for clarity
for p in ax.patches:
    ax.annotate(f'{int(p.get_height())}', 
                (p.get_x() + p.get_width() / 2., p.get_height()), 
                ha='center', va='baseline', 
                fontsize=11, color='black', 
               xytext=(0, 5), textcoords='offset points')

# 5. Add descriptive titles and labels
plt.title('Total Amount of Survivors per Passenger Class', fontsize=14, pad=15)
plt.xlabel('Passenger Class (1 = 1st, 2 = 2nd, 3 = 3rd)', fontsize=12)
plt.ylabel('Number of Survivors', fontsize=12)
```

# Crea una gráfica que muestre la cantidad de sobrevivientes por género (Sex).

```python
ax = sns.countplot(data=survivors_df, x="Sex")
for p in ax.patches:
    print('x', p.get_x())
    print('width', p.get_width())
    print('heigth', p.get_height())
    ax.annotate(p.get_height(), (p.get_x() + p.get_width() / 2., p.get_height()),
               ha="center")
```
